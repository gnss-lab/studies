## Getting data

```python
import os
import requests
from datetime import datetime
from typing import List

def create_directory_structure(dates: List[datetime]) -> None:
    """
    Creates a directory structure based on the given list of dates.
    
    Directories follow the pattern:
    ./data/{year}/{day_of_year}/out/
    
    :param dates: List of date strings in YYYY-MM-DD format.
    """
    os.makedirs(PATH_TO_DATA, exist_ok=True)
    os.makedirs(PATH_TO_WIRED_DATA, exist_ok=True)
    
    for date_obj in dates:
        year = date_obj.strftime("%Y")
        day_of_year = date_obj.strftime("%j")
        
        base_path = f"./data/{year}/{day_of_year}"
        os.makedirs(base_path, exist_ok=True)
        os.makedirs(f"{base_path}/out", exist_ok=True)

def download_data(dates: List[datetime]) -> None:
    """
    Downloads data files into the corresponding directories based on the given dates.
    
    Files are saved in:
    ./data/{year}/{day_of_year}/out/data.h5
    
    :param dates: List of date strings in YYYY-MM-DD format.
    """
    base_url = "https://simurg.space/gen_file?data=obs&date="
    
    for date_obj in dates:
        year = date_obj.strftime("%Y")
        day_of_year = date_obj.strftime("%j")
        
        file_path = f"./data/{year}/{day_of_year}/out/data.h5"
        url = f"{base_url}{date_obj.strftime('%Y-%m-%d')}"
        print(f"Start data download from {url}")
        
        response = requests.get(url, stream=True)
        if response.status_code == 200:
            with open(file_path, "wb") as file:
                for chunk in response.iter_content(chunk_size=8192):
                    file.write(chunk)
            print(f"Downloaded: {file_path}")
        else:
            print(f"Failed to download: {url}")
```
